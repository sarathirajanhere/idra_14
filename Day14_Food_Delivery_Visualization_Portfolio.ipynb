# 🍔 Day 14 — Food Delivery Data Visualization Portfolio
## Matplotlib & Seaborn

An end-to-end visualization portfolio exploring food delivery business performance through time trends, category comparisons, distributions, relationships and correlations.

**Dataset:** Food Delivery Business Performance Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")

FILE="Day14_Food_Delivery_Visualization_Dataset.csv"
df=pd.read_csv(FILE)
df.columns=[c.strip() for c in df.columns]
for c in df.select_dtypes(include="object").columns:
    df[c]=df[c].astype(str).str.strip()
df["Date"]=pd.to_datetime(df["Date"],errors="coerce")

numeric_cols=["Orders","Average_Order_Value","Revenue","Marketing_Spend","Discounts",
              "Avg_Delivery_Minutes","Customer_Rating","Repeat_Customer_Percent"]
for c in numeric_cols:
    df[c]=pd.to_numeric(df[c],errors="coerce")

print("Dataset shape:",df.shape)
display(df.head())
display(df.describe(include="all").T)
```

## 1. 📈 Line Plot — Orders & Revenue Over Time
**Purpose:** Identify changes in business performance over time.

**Interpretation:** The time-series view helps identify periods of stronger and weaker revenue/order activity and provides a basis for operational and promotional planning.

```python
daily=df.groupby("Date").agg(Orders=("Orders","sum"),Revenue=("Revenue","sum"))

fig,ax1=plt.subplots(figsize=(11,5))
ax1.plot(daily.index,daily["Orders"],marker="o",label="Orders")
ax1.set_xlabel("Date"); ax1.set_ylabel("Orders")
ax2=ax1.twinx()
ax2.plot(daily.index,daily["Revenue"],marker="s",label="Revenue")
ax2.set_ylabel("Revenue")
plt.title("Orders and Revenue Over Time")
fig.autofmt_xdate()
plt.show()
```

## 2. 📊 Bar Chart — Revenue by City
**Purpose:** Compare performance across geographic markets.

**Interpretation:** Differences in revenue highlight stronger markets and locations where localized marketing or operational improvements may be required.

```python
city=df.groupby("City").agg(Orders=("Orders","sum"),Revenue=("Revenue","sum")).sort_values("Revenue",ascending=False)
display(city)
plt.figure(figsize=(10,5))
plt.bar(city.index,city["Revenue"])
plt.title("Revenue by City"); plt.xlabel("City"); plt.ylabel("Revenue")
plt.xticks(rotation=30); plt.tight_layout(); plt.show()
```

## 3. 🔵 Scatter Plot — Marketing Spend vs Revenue
**Purpose:** Investigate whether marketing investment is associated with revenue.

**Interpretation:** The direction and strength of the relationship can help assess marketing effectiveness. Association does not by itself establish causation.

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Marketing_Spend"],df["Revenue"],alpha=.7)
plt.title("Marketing Spend vs Revenue")
plt.xlabel("Marketing Spend"); plt.ylabel("Revenue")
plt.tight_layout(); plt.show()
print("Pearson correlation:",df["Marketing_Spend"].corr(df["Revenue"]))
```

## 4. 📉 Histogram — Delivery Time
**Purpose:** Understand the distribution of average delivery times.

**Interpretation:** The histogram shows typical delivery times, spread and unusually long delivery observations that may affect customer experience.

```python
plt.figure(figsize=(8,5))
plt.hist(df["Avg_Delivery_Minutes"].dropna(),bins=20)
plt.title("Distribution of Average Delivery Time")
plt.xlabel("Average Delivery Minutes"); plt.ylabel("Frequency")
plt.tight_layout(); plt.show()
```

## 5. 📦 Box Plot — Customer Ratings by Weather
**Purpose:** Compare rating distributions across weather conditions.

**Interpretation:** Differences in medians, spread and outliers indicate whether weather conditions coincide with changes in customer satisfaction.

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df,x="Weather",y="Customer_Rating")
plt.title("Customer Ratings by Weather")
plt.xlabel("Weather"); plt.ylabel("Customer Rating")
plt.xticks(rotation=20); plt.show()
```

## 6. 🎻 Violin Plot — Customer Ratings by Cuisine
**Purpose:** Compare the shape and spread of rating distributions across cuisines.

**Interpretation:** Violin distributions show where ratings are concentrated and whether some cuisines have wider or more varied customer-rating patterns.

```python
plt.figure(figsize=(9,5))
sns.violinplot(data=df,x="Cuisine",y="Customer_Rating")
plt.title("Customer Rating Distribution by Cuisine")
plt.xlabel("Cuisine"); plt.ylabel("Customer Rating")
plt.xticks(rotation=25); plt.show()
```

## 7. 🧾 Count Plot — Order Channels
**Purpose:** Compare the frequency of different ordering channels.

**Interpretation:** The chart reveals which channels are most represented and provides context for channel-specific performance analysis.

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df,x="Order_Channel")
plt.title("Order Records by Order Channel")
plt.xlabel("Order Channel"); plt.ylabel("Number of Records")
plt.xticks(rotation=20); plt.show()
```

## 8. 🍜 Bar Chart — Orders by Cuisine
**Purpose:** Identify high-volume cuisine categories.

**Interpretation:** Cuisine-level demand concentration can guide menu availability, promotions and inventory planning.

```python
cuisine=df.groupby("Cuisine").agg(Orders=("Orders","sum"),Revenue=("Revenue","sum")).sort_values("Orders",ascending=False)
display(cuisine)
plt.figure(figsize=(10,5))
plt.bar(cuisine.index,cuisine["Orders"])
plt.title("Orders by Cuisine"); plt.xlabel("Cuisine"); plt.ylabel("Total Orders")
plt.xticks(rotation=25); plt.show()
```

## 9. 🌦️ Weather & Performance
**Purpose:** Examine how weather conditions relate to orders, revenue, delivery time and ratings.

```python
weather=df.groupby("Weather").agg(
    Orders=("Orders","sum"),
    Revenue=("Revenue","sum"),
    Avg_Delivery=("Avg_Delivery_Minutes","mean"),
    Avg_Rating=("Customer_Rating","mean")
).sort_values("Revenue",ascending=False)
display(weather.round(2))
```

## 10. 🔥 Correlation Heatmap
**Purpose:** Examine linear relationships among key numerical business metrics.

**Interpretation:** Values closer to +1 or -1 indicate stronger positive or negative linear association, while values near 0 indicate weaker linear association. Correlation should not be interpreted as causation.

```python
corr_cols=["Orders","Average_Order_Value","Revenue","Marketing_Spend","Discounts",
            "Avg_Delivery_Minutes","Customer_Rating","Repeat_Customer_Percent"]
corr=df[corr_cols].corr()

plt.figure(figsize=(10,7))
sns.heatmap(corr,annot=True,fmt=".2f",center=0)
plt.title("Correlation Heatmap — Food Delivery Performance")
plt.tight_layout(); plt.show()
```
